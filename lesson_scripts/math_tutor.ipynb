```python
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic
import sys
sys.path.append("..")  # helpers.py лежит в корне курса
from helpers import add_user_message, chat

client = Anthropic()
model = "claude-sonnet-5-5"
```

```python
# Без system prompt
messages = []
add_user_message(messages, "How do I solve 5x + 2 = 3 for x?")
chat(client, model, messages)
```

```python
# С system prompt
system = """
You are a patient math tutor.
Do not directly answer a student's questions.
Guide them to a solution step by step.
"""

messages = []
add_user_message(messages, "How do I solve 5x + 2 = 3 for x?")
chat(client, model, messages, system=system)
```
