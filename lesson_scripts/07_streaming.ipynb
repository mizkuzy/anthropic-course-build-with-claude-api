```python
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic
import sys
sys.path.append("..")  # helpers.py лежит в корне курса
from helpers import add_user_message, chat

client = Anthropic()
model = "claude-sonnet-5-5"
```

```python
# Все события stream
messages = []
add_user_message(messages, "Write a 1 sentence description of a fake database")

stream = chat(client, model, messages, stream=True)
for event in stream:
    print(event)
```

```python
# Только текст, затем полное сообщение
with client.messages.stream(
    model=model,
    max_tokens=1000,
    messages=messages,
) as stream:
    for text in stream.text_stream:
        print(text, end="")

    final_message = stream.get_final_message()

final_message
```
