```python
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic
import json
import sys
sys.path.append("..")  # helpers.py лежит в корне курса
from helpers import add_user_message, chat, print_results

client = Anthropic()
model = "claude-haiku-5-5"
```

```python
def run_prompt(test_case):
    """Подставляет задачу в prompt и возвращает ответ Claude"""
    prompt = f"""
Please solve the following task:

{test_case["task"]}
"""
    messages = []
    add_user_message(messages, prompt)
    output = chat(client, model, messages)
    return output


def run_test_case(test_case):
    """Вызывает run_prompt и оценивает результат"""
    output = run_prompt(test_case)

    # TODO - Grading
    score = 10

    return {
        "output": output,
        "test_case": test_case,
        "score": score,
    }


def run_eval(dataset):
    """Запускает run_test_case для каждой задачи dataset"""
    results = []
    for test_case in dataset:
        result = run_test_case(test_case)
        results.append(result)
    return results
```

```python
# dataset.json сохранили в уроке 11
with open("dataset.json", "r") as f:
    dataset = json.load(f)

results = run_eval(dataset)
```

```python
print_results(results)
```
