```python
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5-5"
```

```python
# Structured outputs вместо prefill: API возвращает JSON по схеме
import json

response = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[{"role": "user", "content": "Generate a very short event bridge rule as json"}],
    output_config={
        "format": {
            "type": "json_schema",
            "schema": {
                "type": "object",
                "properties": {
                    "source": {"type": "array", "items": {"type": "string"}},
                    "detail-type": {"type": "array", "items": {"type": "string"}},
                },
                "required": ["source", "detail-type"],
                "additionalProperties": False,
            },
        }
    },
)

text = next(block.text for block in response.content if block.type == "text")
rule = json.loads(text)
rule
```
